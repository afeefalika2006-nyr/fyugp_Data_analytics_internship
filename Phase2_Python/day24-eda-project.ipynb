```python
df["Department"] = df["Department"].str.strip()
```

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
```

```python
df["Salary"].fillna(df["Salary"].mean(), inplace=True)
```

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df["PerformanceScore"] = pd.to_numeric(df["PerformanceScore"], errors="coerce")
```

```python
df.isnull().sum()
```

```python
df["PerformanceScore"].fillna(df["PerformanceScore"].mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df["Department"] = df["Department"].str.strip()

df["Department"] = df["Department"].replace("nan","clean")
```

```python
dept_salary = df.groupby("Department")["Salary"].mean()

plt.bar(dept_salary.index, dept_salary.values)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.show()
```

```python
plt.scatter(df["Experience"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
```

```python
plt.hist(df["Salary"], bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.boxplot(df["Salary"])

plt.title("Salary Box Plot")

plt.show()
```

```python
# Observation:-
#1- Finance and IT departments have the highest average salaries.

#2- Most employees earn salaries in the middle salary range.

#3- Employees with higher experience generally earn more.

#4- A few employees have unusually high salaries compared to others.

#5- Employee distribution across cities is fairly balanced.
```
