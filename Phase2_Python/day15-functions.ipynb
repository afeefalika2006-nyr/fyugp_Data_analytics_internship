```python
#pgm1
def cel_to_faren(c):
    f=(c*9/5)+32
    return f

celsius=[0,25,100]
for i in celsius:
    result=cel_to_faren(i)
    print(f"celsius={i}->{result} fahrenheat")
```

```python
#pgm2
def get_grade(score):
    if(score>=90):
        return "Grade A"
    elif(score>=80):
        return "Grade B"
    elif(score>=75):
        return "Grade C"
    elif(score>=50):
        return "Grade D"
    else:
         return "Failed"
              
    print(get_grade(score))
   

s=[76,35,55,83,95]
for i in s:
    result=get_grade(i)
    print(f"Score= {i}->{result}")
```
