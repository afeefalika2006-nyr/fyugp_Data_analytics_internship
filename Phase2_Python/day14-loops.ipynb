```python
#pgm2
scores = [45, 72, 88, 39, 55, 61, 90, 47, 68, 53]
pass_mark=50
for score in scores:
    if(score>=pass_mark):
        print(score,"-pass")
    else:
       print(score,"-fail")
pass_count=0
fail_count=0
for score in scores:
    if score>=pass_mark:
        pass_count += 1
    else:
        fail_count += 1
print(pass_count, "students passed out of 10")
print(fail_count, "students failed out of 10")        
```
