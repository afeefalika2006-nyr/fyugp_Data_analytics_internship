```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("sales_large.csv")
```

```python
plt.scatter(df["Price"], df["Quantity"])

plt.title("Price vs Quantity")
plt.xlabel("Price")
plt.ylabel("Quantity")

plt.show()
```

```python
# Observation:
# The scatter plot shows that high-priced products are generally bought in smaller quantities.
# There is a weak negative relationship between price and quantity.
```

```python
plt.hist(df["Price"])

plt.title("Distribution of Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# Observation:
# The histogram shows that most prices are concentrated in the lower range.
# A few high-priced products appear as outliers.
```
