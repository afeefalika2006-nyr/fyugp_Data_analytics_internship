```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("cleaned_sales_data.csv")

df.head()
```

```python
category_sales = df.groupby("Category")["Total Amount"].sum()

plt.figure(figsize=(8,4))
plt.bar(category_sales.index, category_sales.values, color="skyblue")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.plot(df["Quantity"], marker='o')

plt.title("Quantity Trend")
plt.xlabel("Index")
plt.ylabel("Quantity")

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.hist(df["Unit Price"], bins=10)

plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")

plt.show()
```

```python
region_sales = df.groupby("City")["Total Amount"].sum()

plt.figure(figsize=(8,4))

plt.bar(region_sales.index, region_sales.values)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```
